```python
from langgraph.graph import StateGraph,START,END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
class BlogState(TypedDict):
    topic:str
    outline:str
    blog:str
```

```python
model = ChatGroq(
    model="llama-3.1-8b-instant"  # free model
    
)
```

```python
def create_outline(state:BlogState) ->BlogState:
    #fetch title
    title=state["topic"]
    #call llm from outline
    prompt=f'Genearte a detailed outline for this tipic:{title}'
    outline=model.invoke(prompt).content
    #update state
    state['outline']=outline
    return state
```

```python
def write_blog(state:BlogState) ->BlogState:
    title=state['topic']
    outline=state['outline']
    prompt=f'write a detailed blog for this title :{title} using this following outline:{outline}'
    content=model.invoke(prompt).content
    state['blog']=content
    return state
```

```python
graph=StateGraph(BlogState)
graph.add_node('create_outline',create_outline)
graph.add_node('write_blog',write_blog)

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','write_blog')
graph.add_edge('write_blog',END)

workflow=graph.compile()
```

```python
initial_state={'topic':'The future of AI'}
final_state=workflow.invoke(initial_state)
print(final_state)
```
