```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict,Literal
```

```python
class QuadState(TypedDict):
    a:int
    b:int
    c:int

    eqn:str
    discri:str
    res:str
```

```python
def showeqn(state:QuadState)->QuadState:
    eqn=f'{state["a"]}x2{state["b"]}x{state["c"]}'
    return {"eqn":eqn}
```

```python
def caldiscri(state:QuadState)->QuadState:
    discri=state["b"]**2-(4*state["b"]*state["c"])
    return {"discri":discri}
```

```python
def realroots(state:QuadState)->QuadState:
    root1=(-state["b"] + state["discri"]**0.5)/(2*state["a"])
    root2=(-state["b"] - state["discri"]**0.5)/(2*state["a"])
    res=f'the roots are {root1} and {root2}'
    return {'res':res}
```

```python
def repeatroots(state:QuadState)->QuadState:
    root=-state["b"]/(2*state["a"])
    res=f'the root is {root}'
    return {'res':res}
```

```python
def norealroots(state:QuadState)->QuadState:
    res='there is no real root'
    return {'res':res}
```

```python
#conditional function
def chkcond(state:QuadState) -> Literal['realroots','repeatroots','norealroots']:
    if state['discri']>0:
        return 'realroots'
    elif state['discri']==0:
        return 'repeatroots'
    else:
        return 'norealroots'
    
```

```python
graph=StateGraph(QuadState)
graph.add_node('showeqn',showeqn)
graph.add_node('caldiscri',caldiscri)
graph.add_node('realroots',realroots)
graph.add_node('repeatroots',repeatroots)
graph.add_node('norealroots',norealroots)


graph.add_edge(START,'showeqn')
graph.add_edge('showeqn','caldiscri')
graph.add_conditional_edges('caldiscri',chkcond,['realroots','repeatroots','norealroots'])
graph.add_edge('realroots',END)
graph.add_edge('repeatroots',END)
graph.add_edge('norealroots',END)
workflow=graph.compile()
```

```python
initial_state={
    'a':4,
    'b':-5,
    'c':-4
}
workflow.invoke(initial_state)
```
