```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
```

```python
class Batsmanstate(TypedDict):
    runs:int
    balls:int
    fours:int
    sixes:int
    sr:float
    bpb:float
    bper:float
    summary:str
```

```python
def calsr(state:Batsmanstate)->Batsmanstate:
    runs=state['runs']
    balls=state['balls']
    sr=(runs/balls)*100
    return {'sr':sr}#partial updated state
```

```python
def calbpb(state:Batsmanstate)->Batsmanstate:
    bpb=state['balls']/(state['fours']+state['sixes'])
    
    return {'bpb':bpb}
```

```python
def calbper(state:Batsmanstate)->Batsmanstate:
    bper=((state['fours']*4)+(state['sixes']*6)/state['runs'])*100
    
    return {'bper':bper}
```

```python
def summary(state:Batsmanstate)->Batsmanstate:
    summary=f"""
    Strike rate - {state['sr']}
    Boundary percentage - {state['bper']}
    Balls per boundary - {state['bpb']}
    """
    return {'summary':summary}
```

```python
graph=StateGraph(Batsmanstate)
graph.add_node('calsr',calsr)
graph.add_node('calbpb',calbpb)
graph.add_node('calbper',calbper)
graph.add_node('summary',summary)

graph.add_edge(START,'calsr')
graph.add_edge(START,'calbpb')
graph.add_edge(START,'calbper')
graph.add_edge('calsr','summary')
graph.add_edge('calbpb','summary')
graph.add_edge('calbper','summary')
graph.add_edge('summary',END)

workflow=graph.compile()

```

```python
initial_state={
    'runs':100,
    'balls':50,
    'fours':6,
    'sixes':4
}
final_state=workflow.invoke(initial_state)
print(final_state)
```
